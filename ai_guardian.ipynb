```python
# Create a sample document

sample_text = """
Project Name: AI Guardian

Purpose:
Protect users from harmful AI actions
and unauthorized use of personal information.

Features:
1. AI Risk Detection
2. Permission Guard
3. Human Approval
4. Emergency Stop
5. Personal Image Protection
"""

with open("sample_document.txt", "w") as file:
    file.write(sample_text)

print("Sample document created successfully!")
```

```python

# AI GUARDIAN - STEP 1
# Building a basic AI Agent

def read_document():
    with open("sample_document.txt", "r") as file:
        content = file.read()

    return content


def draft_message():
    message = """
    Hello,

    The AI Guardian project is designed
    to protect users from harmful AI actions.

    Regards,
    AI Guardian Agent
    """

    return message


def request_image_generation():
    return {
        "action": "generate_image",
        "description": "Create a professional portrait"
    }


def run_agent():

    print("\n===== AI GUARDIAN AGENT =====")
    print("1. Read sample document")
    print("2. Draft a message")
    print("3. Request image generation")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nDocument Content:")
        print(read_document())

    elif choice == "2":
        print("\nDraft Message:")
        print(draft_message())

    elif choice == "3":
        print("\nRequested Action:")
        print(request_image_generation())

    elif choice == "4":
        print("Exiting AI Guardian.")

    else:
        print("Invalid choice.")


run_agent()
```

```python

# AI GUARDIAN - STEP 2
# Safety and Permission Engine

def guardian_check(action, user_permission):

    # Block actions if emergency stop is ON
    if emergency_stop:
        return "BLOCKED: Emergency stop is active."

    # Check whether the user has permission
    if not user_permission:
        return "BLOCKED: User permission is missing."

    # Ask approval for sensitive actions
    sensitive_actions = ["send_email", "delete_file"]

    if action in sensitive_actions:
        return "APPROVAL REQUIRED: Ask the user first."

    # Allow safe actions
    return "ALLOWED: Action is safe."


# Emergency stop initially OFF
emergency_stop = False

print("AI Guardian Safety Engine is ready!")
```

```python

# AI GUARDIAN - STEP 3
# Connect Agent with Guardian

# Permissions granted by the user
permissions = {
    "read_file": True,
    "send_email": True,
    "generate_image": True,
    "delete_file": False
}

# Guardian checks every action before execution
def guardian_check(action, approved=False):

    if emergency_stop:
        return "BLOCKED"

    if action not in permissions:
        return "BLOCKED"

    if not permissions[action]:
        return "BLOCKED"

    sensitive_actions = ["send_email", "delete_file"]

    if action in sensitive_actions and not approved:
        return "APPROVAL REQUIRED"

    return "ALLOWED"


# Execute only after Guardian allows the action
def execute_action(action):

    if action == "read_file":
        return read_document()

    elif action == "send_email":
        return "Demo: Email draft prepared. No real email sent."

    elif action == "generate_image":
        return "Demo: Image generation tool requested."

    elif action == "delete_file":
        return "Demo: Delete operation requested."

    return "Unknown action"


# AI Agent requests an action
def ai_agent(action, approved=False):

    print("\nAI Agent requests:", action)

    decision = guardian_check(action, approved)

    print("Guardian decision:", decision)

    if decision == "ALLOWED":
        result = execute_action(action)
        print("\nTask result:")
        print(result)

    elif decision == "APPROVAL REQUIRED":
        print("Guardian: User approval is required.")

    else:
        print("Guardian: Action stopped for safety.")


print("AI Agent connected to AI Guardian!")
```

```python

# AI GUARDIAN - STEP 4
# Personal Photo Protection

# Permissions set by the photo owner
photo_permissions = {
    "professional_portrait": True,
    "background_change": True,
    "clothing_change": True,
    "harmful_edit": False,
    "share_externally": False
}


def photo_guardian(action):

    print("\nPhoto request:", action)

    # Check whether the action exists
    if action not in photo_permissions:
        print("Guardian: BLOCKED")
        print("Reason: Unknown or unauthorized action.")
        return "BLOCKED"

    # Check the owner's permission
    if photo_permissions[action] == False:
        print("Guardian: BLOCKED")
        print("Reason: The photo owner has not permitted this use.")
        return "BLOCKED"

    # Allow permitted actions
    print("Guardian: ALLOWED")
    print("Reason: The photo owner has permitted this use.")

    return "ALLOWED"


print("Personal Photo Protection is ready!")
```

```python

# Simulated image-generation tool

def image_generation_tool(action):

    print("\nImage tool received the request.")

    if action == "professional_portrait":
        print("Generating a professional portrait...")

    elif action == "background_change":
        print("Changing the photo background...")

    elif action == "clothing_change":
        print("Applying the permitted clothing edit...")

    else:
        print("This action is not supported by the demo.")


# Guardian controls access to the image tool
def protected_image_request(action):

    decision = photo_guardian(action)

    if decision == "ALLOWED":
        image_generation_tool(action)

    else:
        print("Image tool was NOT called.")
        print("Guardian prevented unauthorized use.")


print("Protected image-generation system is ready!")
```

```python

# AI GUARDIAN - STEP 6
# Combined safety engine and activity log

from datetime import datetime

# Activity history
guardian_logs = []

# Emergency stop
emergency_stop = False

# Permissions
agent_permissions = {
    "read_file": True,
    "draft_message": True,
    "send_email": True,
    "delete_file": False
}


def guardian_agent(action, approved=False):

    global emergency_stop

    # Emergency stop check
    if emergency_stop:
        decision = "BLOCKED"
        reason = "Emergency stop is active."

    # Permission check
    elif action not in agent_permissions:
        decision = "BLOCKED"
        reason = "Unknown action."

    elif not agent_permissions[action]:
        decision = "BLOCKED"
        reason = "Action is not permitted."

    # Human approval check
    elif action == "send_email" and not approved:
        decision = "APPROVAL REQUIRED"
        reason = "User approval is required."

    else:
        decision = "ALLOWED"
        reason = "Action is permitted."

    # Record every decision
    guardian_logs.append({
        "time": datetime.now().strftime("%H:%M:%S"),
        "action": action,
        "decision": decision,
        "reason": reason
    })

    # Execute only allowed actions
    if decision == "ALLOWED":

        if action == "read_file":
            result = read_document()

        elif action == "draft_message":
            result = draft_message()

        elif action == "send_email":
            result = "Demo email prepared. No real email sent."

        else:
            result = "Action completed in demo mode."

    else:
        result = reason

    return decision, result


print("Combined AI Guardian system is ready!")
```

```python

import gradio as gr
import pandas as pd
from datetime import datetime

# Guardian dashboard status
guardian_active = True
guardian_dashboard_logs = []

# Main Guardian action function
def run_guardian_dashboard(action, approved):
    global guardian_active

    if not guardian_active:
        decision = "BLOCKED"
        result = "Emergency Stop is active."

    elif action == "Read File":
        decision, result = guardian_agent("read_file")

    elif action == "Draft Message":
        decision, result = guardian_agent("draft_message")

    elif action == "Send Email":
        decision, result = guardian_agent(
            "send_email", approved=approved
        )

    elif action == "Delete File":
        decision, result = guardian_agent("delete_file")

    else:
        decision = "BLOCKED"
        result = "Unknown action."

    guardian_dashboard_logs.append({
        "Time": datetime.now().strftime("%H:%M:%S"),
        "Action": action,
        "Decision": decision,
        "Result": str(result)
    })

    log_df = pd.DataFrame(guardian_dashboard_logs)

    return decision, str(result), log_df


# Emergency Stop function
def activate_emergency_stop():
    global guardian_active, emergency_stop

    guardian_active = False
    emergency_stop = True

    return "EMERGENCY STOP ACTIVATED! All controlled actions blocked."


# Reset Guardian function
def reset_guardian():
    global guardian_active, emergency_stop

    guardian_active = True
    emergency_stop = False

    return "Guardian restarted. Safety checks are active."


# Create dashboard
with gr.Blocks(title="AI Guardian Safety Dashboard") as guardian_demo:

    gr.Markdown("# AI Guardian")
    gr.Markdown(
        "AI that works for you, while you stay in control."
    )

    with gr.Row():
        action = gr.Dropdown(
            choices=[
                "Read File",
                "Draft Message",
                "Send Email",
                "Delete File"
            ],
            label="Select AI Agent Action",
            value="Read File"
        )

        approved = gr.Checkbox(
            label="I approve sending the demo email",
            value=False
        )

    run_button = gr.Button("Run Guardian", variant="primary")

    decision_output = gr.Textbox(label="Guardian Decision")
    result_output = gr.Textbox(label="Action Result")

    run_button.click(
        fn=run_guardian_dashboard,
        inputs=[action, approved],
        outputs=[
            decision_output,
            result_output,
            gr.State()
        ]
    )

    gr.Markdown("## Emergency Controls")

    with gr.Row():
        stop_button = gr.Button(
            "EMERGENCY STOP",
            variant="stop"
        )
        reset_button = gr.Button("Restart Guardian")

    emergency_status = gr.Textbox(label="System Status")

    stop_button.click(
        fn=activate_emergency_stop,
        outputs=emergency_status
    )

    reset_button.click(
        fn=reset_guardian,
        outputs=emergency_status
    )

    gr.Markdown("## Audit Log")

    log_output = gr.Dataframe(
        headers=["Time", "Action", "Decision", "Result"],
        label="Guardian Activity"
    )

    run_button.click(
        fn=run_guardian_dashboard,
        inputs=[action, approved],
        outputs=[
            decision_output,
            result_output,
            log_output
        ]
    )

guardian_demo.launch(share=True)
```

```python

import gradio as gr
import pandas as pd
from datetime import datetime

# Shared dashboard state
guardian_active = True
emergency_stop = False
combined_logs = []


# Function to save activity in the common audit log
def save_log(action, decision, reason):
    combined_logs.append({
        "Time": datetime.now().strftime("%H:%M:%S"),
        "Action": action,
        "Decision": decision,
        "Reason": str(reason)
    })


# AI Agent Safety
def run_agent_dashboard(action, approved):
    if emergency_stop or not guardian_active:
        decision = "BLOCKED"
        result = "Emergency Stop is active."

    else:
        action_map = {
            "Read File": "read_file",
            "Draft Message": "draft_message",
            "Send Email": "send_email",
            "Delete File": "delete_file"
        }

        decision, result = guardian_agent(
            action_map[action],
            approved=approved
        )

    save_log(action, decision, result)

    return (
        decision,
        str(result),
        pd.DataFrame(combined_logs)
    )


# Personal Photo Protection
def run_photo_dashboard(image, action):
    if emergency_stop or not guardian_active:
        decision = "BLOCKED"
        result = "Emergency Stop is active."
        preview = None

    elif image is None:
        decision = "BLOCKED"
        result = "Please upload a sample photo."
        preview = None

    elif not photo_permissions.get(action, False):
        decision = "BLOCKED"
        result = "Permission denied for " + action
        preview = None

    else:
        decision = "ALLOWED"
        result = "Permission verified for " + action
        preview = image

    save_log("Photo: " + str(action), decision, result)

    return (
        decision,
        result,
        preview,
        pd.DataFrame(combined_logs)
    )


# Emergency Stop
def stop_guardian():
    global guardian_active, emergency_stop

    guardian_active = False
    emergency_stop = True

    save_log(
        "Emergency Stop",
        "ACTIVATED",
        "Controlled agent and photo actions paused."
    )

    return (
        "EMERGENCY STOP ACTIVATED!",
        pd.DataFrame(combined_logs)
    )


# Restart Guardian
def restart_guardian():
    global guardian_active, emergency_stop

    guardian_active = True
    emergency_stop = False

    save_log(
        "Guardian Restart",
        "RESTARTED",
        "Safety checks are active again."
    )

    return (
        "Guardian restarted. Safety checks are active.",
        pd.DataFrame(combined_logs)
    )


# Create one combined dashboard
with gr.Blocks(title="AI Guardian - Unified Safety Dashboard") as final_demo:

    gr.Markdown("# AI Guardian")
    gr.Markdown(
        "AI that works for you, while you stay in control."
    )

    gr.Markdown(
        "## Module 1: AI Agent Safety"
    )

    agent_action = gr.Dropdown(
        choices=[
            "Read File",
            "Draft Message",
            "Send Email",
            "Delete File"
        ],
        value="Read File",
        label="Select Agent Action"
    )

    agent_approval = gr.Checkbox(
        label="I approve sending the demo email",
        value=False
    )

    agent_button = gr.Button(
        "Run AI Agent",
        variant="primary"
    )

    agent_decision = gr.Textbox(
        label="Guardian Decision"
    )

    agent_result = gr.Textbox(
        label="Agent Result"
    )

    gr.Markdown("## Module 2: Personal Photo Protection")

    photo_input = gr.Image(
        type="pil",
        label="Upload Sample Photo"
    )

    photo_action = gr.Dropdown(
        choices=[
            "professional_portrait",
            "background_change",
            "clothing_change",
            "harmful_edit",
            "share_externally"
        ],
        value="background_change",
        label="Intended Photo Use"
    )

    photo_button = gr.Button(
        "Check Photo Permission",
        variant="primary"
    )

    photo_decision = gr.Textbox(
        label="Photo Guardian Decision"
    )

    photo_result = gr.Textbox(
        label="Photo Protection Result"
    )

    photo_preview = gr.Image(
        label="Approved Photo Preview"
    )

    gr.Markdown("## Module 3: Shared Emergency Controls")

    with gr.Row():
        stop_button = gr.Button(
            "EMERGENCY STOP",
            variant="stop"
        )

        restart_button = gr.Button(
            "Restart Guardian"
        )

    system_status = gr.Textbox(
        label="System Status"
    )

    gr.Markdown("## Module 4: Combined Audit Log")

    audit_output = gr.Dataframe(
        headers=[
            "Time",
            "Action",
            "Decision",
            "Reason"
        ],
        label="Guardian Activity Log"
    )

    # Connect agent button
    agent_button.click(
        fn=run_agent_dashboard,
        inputs=[agent_action, agent_approval],
        outputs=[
            agent_decision,
            agent_result,
            audit_output
        ]
    )

    # Connect photo button
    photo_button.click(
        fn=run_photo_dashboard,
        inputs=[photo_input, photo_action],
        outputs=[
            photo_decision,
            photo_result,
            photo_preview,
            audit_output
        ]
    )

    # Connect emergency controls
    stop_button.click(
        fn=stop_guardian,
        outputs=[
            system_status,
            audit_output
        ]
    )

    restart_button.click(
        fn=restart_guardian,
        outputs=[
            system_status,
            audit_output
        ]
    )


# Launch unified dashboard
final_demo.launch(share=True)
```
